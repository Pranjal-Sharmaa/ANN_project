```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , LabelEncoder , OneHotEncoder
import pickle
```

```python
#load the datset
data=pd.read_csv(r'D:\ANN_project\Churn_Modelling.csv')
data.head()
```

```python
##preprocess the data 
##droping irrelevent columns
data=data.drop(['RowNumber','CustomerId','Surname'],axis=1)
data.head()
```

```python
## encode categorical variable
label_encoder_gender=LabelEncoder()
data['Gender']=label_encoder_gender.fit_transform(data['Gender'])
data.head()
```

```python
data['Geography'].unique()
```

```python
##onehot encoder 'geography'
onehot_encoder_geography=OneHotEncoder()
geo_encoder=onehot_encoder_geography.fit_transform(data[['Geography']])
geo_encoder
```

```python
geo_encoder.toarray()
```

```python
data['Geography'].unique()
```

```python
geo_encoder_df=pd.DataFrame(geo_encoder.toarray(),columns=onehot_encoder_geography.get_feature_names_out(['Geography']))
geo_encoder_df
```

```python
data.head()
```

```python
#combine all columns with original data
data = data.drop(['Geography'],axis=1)
data = pd.concat([data, geo_encoder_df],axis=1)
```

```python
data.head()
```

```python
#save the encoder and scaler
with open ('label_encoder_gender.pkl','wb') as file:
    pickle.dump(label_encoder_gender,file)

with open ('onehot_encoder_geography.pkl','wb') as file:
    pickle.dump(onehot_encoder_geography,file)    
```

```python
#decide the dataset into dependent and independent
x=data.drop('Exited',axis=1)
y=data['Exited']
```

```python
#train test splt
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

#sacle the features
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
x_train
```

```python
with open ('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)  
```

ANN Implementation


```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks  import EarlyStopping , TensorBoard
```

```python
#build ANN model
model=Sequential([
    Dense(64,activation='relu',input_shape=(x_train.shape[1],)),#HL1
    Dense(32,activation='relu'),#HL2
    Dense(1,activation='sigmoid')# output layuer
])
```

```python
model.summary()
```

```python
import tensorflow
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
#compile the model
model.compile(optimizer=opt,loss="binary_crossentropy",metrics=['accuracy'])
```

```python
## set up tensorboard
import datetime
log_dir="logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
##setup earlystopping
early_stopping_callback=EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
```

```python
## train the model
history= model.fit(
    x_train,y_train,validation_data=(x_test,y_test),epochs=100,
    callbacks=[tensorflow_callback,early_stopping_callback]
)
```

```python
model.save('model.h5')
```
